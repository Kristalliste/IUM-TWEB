This dataset contains over 940.000 movies. For each movie it displays:

* `tagline` poster tagline;

```python
movies_df = pd.read_csv('../../Data/movies.csv')
```

```python
movies_df.shape
```

### Data Types


```python
movies_df.dtypes
```

We convert:
* name of type `object` into `string`;
* date of type `float` into `date`;
* tagline of type `object` into `string`;
* description of type `object` into `string`;

```python
movies_df['date'] = pd.to_datetime(movies_df['date'], format='%Y', errors='coerce')
movies_df['name'] = movies_df['name'].astype('string')
movies_df['tagline'] = movies_df['tagline'].astype('string')
movies_df['description'] = movies_df['description'].astype('string')
```

Results of the conversion:

```python
movies_df.dtypes
```

Next, we will be converting all `name`, `tagline` and `description` entries to lower case and eliminating placeholder names or symbols. All strings that get emptied will be turned `null`, to be eliminated afterwards.

```python
# Define unwanted patterns and characters
patterns_to_remove = r'^-+$|^-placeholder-$|^-Placeholder-$'
chars_to_remove = r"[.,\-#\"]"

# Select columns to clean
cols_to_clean = ['name', 'tagline', 'description']

movies_df[cols_to_clean] = (
    movies_df[cols_to_clean]
    .apply(lambda col: col.str.strip().str.lower()
           .str.replace(chars_to_remove, '', regex=True)
           .str.replace(r'\.$', '', regex=True)  # removes dots at end of string
           .str.replace(patterns_to_remove, '', regex=True)
    .replace('', np.nan))  # Convert empty strings to NaN
)
```

```python
movies_df.isna().sum()
```

We decided to drop rows where the `name` entry is _null_.

```python
movies_df = movies_df.dropna(subset=['name'], how='any')
```

Result of the operation.

```python
movies_df.isna().sum()
```
