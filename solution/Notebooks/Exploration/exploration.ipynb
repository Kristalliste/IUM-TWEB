## Movies Dataframe
This dataset contains over 940.000 movies. For each movie it displays:
* `id` unique value that identify a movie;
* `name` of the movie;
* `date` of when the movie was released;
* `tagline` poster tagline;
* `description` short description;
* `minute` duration of the movie;
* `rating` of the movie.



```python
movies_df = pd.read_csv('../../Data/movies.csv')
```

```python
movies_df.shape
```

```python
movies_df.head()
```

### Data Types


```python
movies_df.dtypes
```

We convert:
* name of type `object` into `string`;
* date of type `float` into `date`;
* tagline of type `object` into `string`;
* description of type `object` into `string`;

```python
movies_df['date'] = pd.to_datetime(movies_df['date'], format='%Y', errors='coerce')
movies_df['name'] = movies_df['name'].astype('string')
movies_df['tagline'] = movies_df['tagline'].astype('string')
movies_df['description'] = movies_df['description'].astype('string')
```

Result of the conversion.

```python
movies_df.dtypes
```

### Checking for _null_ values


```python
movies_df.isna().sum()
```

We decided to drop che rows where the name is null.

```python
movies_df = movies_df.dropna(subset=['name'], how='any')
```

Result of the operation

```python
movies_df.isna().sum()
```

### Checking for _duplicated_ values

```python
movies_df.duplicated().sum()
```

There are no duplicated rows in the dataframe. Let's check for specific columns.

```python
movies_df.duplicated(subset=['id']).sum()
```

The id is unique in the whole dataframe.

```python
print(movies_df.duplicated(subset=['name'], keep=False).sum())
```

We found more than 200.000 duplicated names. Let's analyze them.

```python
valori_duplicati = movies_df[movies_df.duplicated(subset=['name'], keep=False)]
valori_duplicati = valori_duplicati.sort_values(by='name')
valori_duplicati
```

We decided to remove the rows where there are multiple duplicated:
* same `name`, `date`, `tagline`, `description`, `minute`, `rating`;
* same `name`, but null `date`, `tagline`, `description`, `minute`, `rating`;
* same `name`, `date`, `tagline`, but null `description`, `minute`, `rating`;

```python
# 1. Rimuovi i duplicati considerando tutte le colonne principali
movies_df = movies_df.drop_duplicates(subset=['name', 'date', 'tagline', 'description', 'minute', 'rating'], keep=False)

# 2. Tieni solo i record dove 'date' è null e togli duplicati basandoti su altre colonne
movies_df = movies_df[movies_df['date'].isna()]
movies_df = movies_df.drop_duplicates(subset=['name', 'tagline', 'description', 'minute', 'rating'], keep=False)

# 3. Poi ancora, su quelli dove 'description' è null
movies_df = movies_df[movies_df['description'].isna()]
movies_df = movies_df.drop_duplicates(subset=['name', 'date', 'tagline', 'minute', 'rating'], keep=False)
```

### Checking for _rating range_

```python
movies_df[(movies_df['rating'] > 5) | (movies_df['rating'] < 0)].shape[0]
```

All the rows have a rating that is in the range of 0 to 5.
